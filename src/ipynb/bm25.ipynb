```python
# 英文/数字串 或 单个汉字
import re


_TOKEN_RE = re.compile(r"[A-Za-z0-9_./:@-]+|[\u4e00-\u9fff]")


def tokenize(text: str) -> list[str]:
    return [t.lower() for t in _TOKEN_RE.findall(text)]
```

```python
tokenize("""
工号: XY004
姓名: 赵宁
部门: 信息安全组
岗位: 安全工程师
邮箱: zhaoning@xingyun-example.com
分机: 8003
办公地: 上海
""")
```
